CEP ML Preprocessing Pipeline — Entry Point

Runs all four steps in order:
  1. Preprocess  — .webm → 16kHz WAV, normalize, trim silence, rename (e.g. eng_calm_001.wav)
  2. Augment     — balance minority classes (pitch / stretch / noise)
  3. Metadata    — write processed_metadata.csv
  4. EDA         — statistics tables + acoustic charts

Usage:
    python main.py

Requirements:
    pip install -r requirements.txt
    ffmpeg must be on PATH (needed for .webm decoding)

```python
!pip install pydub openpyxl

from pipeline import preprocess, augment, metadata, eda
from pipeline.config import LABEL_MAP, METADATA_FILE, EDA_DIR
```

```python
def main():

    print("PREPROCESSING PIPELINE")
    print("  Source : DATASET/raw dataset/<language>/<emotion>/*.webm")
    print("  Output : DATASET/processed dataset/  |  processed_metadata.xlsx  |  milestone 02 shots/eda/")

    print("\n  Label encoding:")
    for emo, lbl in LABEL_MAP.items():
        print(f"    {emo:<12} → {lbl}")

    # Run pipeline steps
    
    # Step 1: .webm -> 16kHz WAV, peak-normalize, trim silence
    preprocess.run()

    # Step 2: Augment minority classes until all match the largest class
    augment.run()

    # Step 3: Scan processed dataset/ and write processed_metadata.xlsx
    df = metadata.run()

    # Step 4: Print stats tables and save EDA charts
    eda.run(df)

    print("Pipeline complete!")
    print(f"  EXCEL : {METADATA_FILE}")
    print(f"  EDA : {EDA_DIR}")
```

```python
main()
```
